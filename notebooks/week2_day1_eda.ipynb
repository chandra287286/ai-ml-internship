```python
# Week 2 – Day 1: Exploratory Data Analysis

## Initial EDA Summary

1. The dataset contains 10 records and 4 columns covering question categories, system prompts, and test questions.
2. The `category` column will be analyzed to understand the distribution of question types.
3. The `id` column is an identifier and will not be treated as a meaningful numerical feature.
4. Correlation analysis will be limited to meaningful numerical features; the current dataset has no suitable numerical features for correlation analysis.
5. Missing values will be checked across all columns, and the category distribution and text patterns will be examined.

*These findings will be verified and updated after executing the notebook.*
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

DATASET_PATH = Path("../data/sample_dataset.csv")

df = pd.read_csv(DATASET_PATH)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Information")
print("-" * 40)

df.info()

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Shape:")
print(df.shape)
```

```python
print("Numerical Statistics:")
display(df.describe())

print("\nStatistics for All Columns:")
display(df.describe(include="all"))
```

```python
category_counts = df["category"].value_counts()

print("Category Distribution:")
display(category_counts)

plt.figure(figsize=(10, 5))

category_counts.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Distribution of Question Categories")
plt.xlabel("Category")
plt.ylabel("Number of Questions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Exclude ID because it is only a record identifier.
numeric_features = df.select_dtypes(include=np.number).columns.tolist()
numeric_features = [
    col for col in numeric_features if col != "id"
]

if len(numeric_features) >= 2:
    correlation_matrix = df[numeric_features].corr()

    pairs = (
        correlation_matrix.where(
            np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
        )
        .stack()
        .sort_values(key=abs, ascending=False)
    )

    print("Top 3 Feature Correlations:")
    display(pairs.head(3))

    plt.figure(figsize=(8, 5))
    sns.heatmap(
        correlation_matrix,
        annot=True,
        cmap="coolwarm",
        center=0
    )
    plt.title("Feature Correlation Heatmap")
    plt.show()

else:
    print("Correlation analysis is not applicable.")
    print("Reason: No meaningful numerical features are available.")
```

```python
missing_counts = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing Values": missing_counts,
    "Missing Percentage": missing_percentage
})

print("Missing Value Report:")
display(missing_report)

high_missing_columns = missing_report[
    missing_report["Missing Percentage"] > 5
]

print("\nColumns with more than 5% missing values:")
display(high_missing_columns)

if high_missing_columns.empty:
    print("No columns have more than 5% missing values.")
```

```python
print("Duplicate Records:", df.duplicated().sum())

print("\nUnique Values Per Column:")
display(df.nunique())
```

```python
print("========== FINAL EDA OBSERVATIONS ==========")

print("Dataset shape:", df.shape)

print("\nCategory distribution:")
print(df["category"].value_counts())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())

print("\nNumeric columns:", df.select_dtypes(include=np.number).columns.tolist())

print("\nEDA completed successfully!")
```
